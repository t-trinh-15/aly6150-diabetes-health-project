```python
# MEPS Table C: stratified by high_spender 

# Table C — MEPS, stratified by high_spender flag (top 25% vs. bottom 75%)
# This sets up the predictive question: who are the top spenders, and how
# do they differ from everyone else?

# We need a labeled version of the high_spender flag for nicer column headers
meps_for_table_c = meps.copy()
meps_for_table_c["spender_group"] = meps_for_table_c["high_spender"].map({
    0: "Bottom 75%",
    1: "Top 25%",
})

# Build a variable spec that excludes high_spender itself (it's the grouping var)
# and the dollar variables (since high_spender is defined from total_medical_expense,
# including it would be circular for the test). Keep them in for descriptive
# context but they will look mechanically extreme.
from src.descriptive_tables import MEPS_VARS

# Filter out the variables that would be circular / uninformative as predictors
MEPS_TABLE_C_VARS = {
    k: v for k, v in MEPS_VARS.items()
    if k != "high_spender"  # the grouping variable
}

meps_table_c = build_table_b(
    meps_for_table_c,
    MEPS_TABLE_C_VARS,
    group_col="spender_group",
    group_label="Spender group",
    dataset_label="MEPS HC-251 (2023) — by spender group",
)

out_path = OUTPUT_TABLES / "meps_table_c_by_spender.xlsx"
save_table(meps_table_c, out_path, sheet_name="By spender group")
print(f"Saved: {out_path}")

meps_table_c
```

```python
# Spot check of the outcome-exposure relationship

# Does insurance type predict total medical spending (MEPS)?
# Does uninsured status predict missed care (BRFSS)?

print("=" * 60)
print("OUTCOME–EXPOSURE SPOT CHECKS")
print("=" * 60)

# MEPS: median total expense by insurance type
print("\nMEPS — Median total medical expense by insurance type:")
print(meps.groupby("insurance_label")["total_medical_expense"].agg(["count", "median", "mean"]).round(0).to_string())

# BRFSS: missed care rate by uninsured status
print("\nBRFSS — Missed-care-cost rate by insurance status:")
print(brfss.groupby("uninsured")["missed_care_cost"].agg(["count", "mean"]).round(3).to_string())
print("  (uninsured: 0 = covered, 1 = uninsured)")
```
